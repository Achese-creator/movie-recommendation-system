```python
import pandas as pd
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
TMDB_RAW_DIR = PROJECT_ROOT / "data" / "raw" / "tmdb"

TMDB_MOVIES_PATH = TMDB_RAW_DIR / "tmdb_5000_movies.csv"
TMDB_CREDITS_PATH = TMDB_RAW_DIR / "tmdb_5000_credits.csv"
```

```python
print(TMDB_MOVIES_PATH)
print(TMDB_CREDITS_PATH)
```

```python
tmdb_movies = pd.read_csv(TMDB_MOVIES_PATH)
tmdb_credits = pd.read_csv(TMDB_CREDITS_PATH)
```

```python
print("TMDB movies shape:", tmdb_movies.shape)
print("TMDB credits shape:", tmdb_credits.shape)
```

```python
print("TMDB movies columns:")
print(tmdb_movies.columns)

print("\nTMDB credits columns:")
print(tmdb_credits.columns)
```

```python
print("TMDB movies data types:")
print(tmdb_movies.dtypes)

print("\nTMDB credits data types:")
print(tmdb_credits.dtypes)
```

```python
tmdb_movies.isna().sum().sort_values(ascending=False)
```

```python
tmdb_credits.isna().sum().sort_values(ascending=False)
```

```python
missing_percent_movies = tmdb_movies.isna().mean().sort_values(ascending=False) * 100
missing_percent_movies
```

```python
print("Duplicate rows in TMDB movies:", tmdb_movies.duplicated().sum())
print("Duplicate rows in TMDB credits:", tmdb_credits.duplicated().sum())
```

```python
print("Duplicate movie IDs in TMDB movies:", tmdb_movies["id"].duplicated().sum())
print("Duplicate movie IDs in TMDB credits:", tmdb_credits["movie_id"].duplicated().sum())
```
